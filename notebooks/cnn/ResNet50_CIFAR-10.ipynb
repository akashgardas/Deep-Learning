## Imports

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

## Data

### Load and Preprocess Data

```python
(x_train, y_train), (x_test, y_test) = cifar10.load_data()

# Resize to 224x224
x_train = tf.image.resize(x_train, [224,224])
x_test = tf.image.resize(x_test, [224,224])

# Convert to numpy arrays
x_train = x_train.numpy()
x_test = x_test.numpy()
```

```python
# Preprocess for ResNet50 (mean subtraction, scaling)
from tensorflow.keras.applications.resnet50 import preprocess_input
x_train = preprocess_input(x_train)
x_test = preprocess_input(x_test)
```

```python
# One-hot encode labels
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

## Build Model

### Load Pretrained ResNet50 (Feature Extractor )

```python
datagen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True
)
datagen.fit(x_train)
```

### Build Transfer Learning Model

```python
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False  # freeze base

model = Sequential([
    base_model,
    GlobalAveragePooling2D(),
    Dense(256, activation='relu'),
    Dropout(0.5),
    Dense(10, activation='softmax')
])

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])
model.summary()
```

## Train Model

```python
history = model.fit(datagen.flow(x_train, y_train, batch_size=64),
                    epochs=10, validation_data=(x_test, y_test))
```

## Evaluate Model

```python
base_model.trainable = True
for layer in base_model.layers[:-20]:  # keep most layers frozen
    layer.trainable = False

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1e-5),
              loss='categorical_crossentropy',
              metrics=['accuracy'])

history_ft = model.fit(datagen.flow(x_train, y_train, batch_size=64),
                       epochs=10, validation_data=(x_test, y_test))
```

```python
loss, acc = model.evaluate(x_test, y_test, verbose=0)
print(f"Final Test Accuracy: {acc:.4f}")
```
